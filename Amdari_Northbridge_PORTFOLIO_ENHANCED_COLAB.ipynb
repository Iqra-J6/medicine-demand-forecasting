# SECTION 11 — Introduce Time Series

A **Time Series** is a sequence of data points recorded at regular successive intervals over time.

### The Three Fundamental Components:
1. **Trend**: The long-term directional movement over years (e.g., growing patient population increasing baseline medicine demand).
2. **Seasonality**: Patterns that repeat at predictable, cyclical calendar intervals (e.g., winter respiratory illnesses surging every December/January).
3. **Noise / Randomness**: Irregular, unpredictable short-term fluctuations.

```
       [ Historical Data (Past) ]                       [ Future Horizon ]
2021    2022    2023    2024                             2025
 ───►    ───►    ───►    ───►                             ───►
[Trend + Winter Seasonality + Hospital Activity]  ===>  [Forecast Next Month]
```

### The Concept of a "Lag":
* In time series, what happened in the past strongly influences what will happen next.
* A **Lag-1** feature represents: *"What was demand last month?"*
* A **Lag-12** feature represents: *"What was demand in the exact same month last year?"*

# SECTION 14 — Train/Test Split

### Why Random Splitting is Dangerous for Time Series:
In standard tabular machine learning (e.g. predicting patient disease risk), you can randomly shuffle rows.  
**In time series forecasting, random shuffling is strictly forbidden!**

If you randomly shuffle time series data:
* Your model trains on data from December 2024 to "predict" November 2024.
* This is **Data Leakage** (peeking into the future).
* The model will look artificially accurate during training, but will completely fail in production when real future data arrives.

### The Professional Approach: Chronological Splitting
We train strictly on the **Past** and evaluate on the **Future**:
* **Training Set**: 2021-01-01 to 2024-12-01 (4 years / 48 months)
* **Test Set**: 2025-01-01 to 2025-12-01 (1 full year / 12 months)

```
[================= TRAINING SET (2021 - 2024) =================] [== TEST SET (2025) ==]
◄──────────────────────── Past History ─────────────────────────► ◄── Future Evaluation ──►
```

# SECTION 16 & 17 — SARIMAX: Statistical Forecasting

### What is SARIMAX?
**SARIMAX** stands for:
* **S**easonal: Models recurring annual cycles (e.g. 12-month winter surges).
* **A**uto**R**egressive (p): Uses weighted combinations of past values.
* **I**ntegrated (d): Differences data to stabilize trends.
* **M**oving **A**verage (q): Adjusts for past forecast error shocks.
* **eX**ogenous variables (X): Incorporates external explanatory drivers (such as `Admissions` and `Emergency_Admissions`).

### Why is SARIMAX Loved in Healthcare?
It is mathematically rigorous, handles seasonality natively, and allows clinicians to ask: *"If hospital admissions rise by 15%, how much extra medicine will we need?"*

````python
# Step 18.1: Fit SARIMAX Model with Hospital Activity as Exogenous Covariates
from statsmodels.tsa.statespace.sarimax import SARIMAX

exog_features = ['Admissions', 'Emergency_Admissions']

if train_data is not None and len(train_data) > 0:
    sarima_model = SARIMAX(
        train_data['Quantity_Dispensed'],
        exog=train_data[exog_features],
        order=(1, 1, 1),
        seasonal_order=(1, 1, 0, 12),
        enforce_stationarity=False,
        enforce_invertibility=False
    ).fit(disp=False)

    print("SARIMAX model training complete.")
else:
    print("Please complete the train/test split in Task 6 first.")
````

### What are we doing?
We are fitting a SARIMAX model on historical training data and generating predictions for all 12 months of 2025.

### Your Turn:
Call `.forecast()` on `sarima_model` with `steps=len(test_data)` and `exog=test_data[exog_features]`.

````python
# ==============================================================================
# YOUR TASK 7
# Generate SARIMAX Forecast for 2025
# ==============================================================================

# TODO: Generate 12 forecast steps using sarima_model.forecast(...)
sarimax_forecast = sarima_model.forecast(steps=len(test_data), exog=test_data[exog_features])

if sarimax_forecast is not None:
    sarimax_forecast = pd.Series(sarimax_forecast.values, index=test_data.index)
    print("SARIMAX 2025 Forecast generated:")
    display(pd.DataFrame({
        'Actual_Demand': test_data['Quantity_Dispensed'],
        'SARIMAX_Forecast': sarimax_forecast.round(1)
    }))

# Run Grader
gu.check_task_7(sarimax_forecast)
````

# SECTION 19 & 20 — XGBoost: Machine Learning Forecasting

### Why Compare with XGBoost?
* While SARIMAX is a classical econometric time-series model, **XGBoost** (Extreme Gradient Boosting) is one of the most powerful machine learning algorithms for tabular data.
* XGBoost builds an ensemble of decision trees that learn non-linear relationships across our engineered lag features, calendar months, and hospital activity drivers.

Let's train an XGBoost model and compare both approaches!

````python
# Step 19.1: Train the XGBoost Regressor
import xgboost as xgb

# Define feature matrix
features = [
    'Demand_Lag_1', 'Demand_Lag_2', 'Demand_Lag_3', 'Demand_Lag_12',
    'Rolling_3_Mean', 'Month', 'Quarter',
    'Admissions', 'Emergency_Admissions', 'Bed_Occupancy'
]

if train_data is not None and len(train_data) > 0 and 'Demand_Lag_1' in train_data.columns and train_data['Demand_Lag_1'].notnull().any():
    X_train = train_data[features]
    y_train = train_data['Quantity_Dispensed']
    X_test = test_data[features]
    y_test = test_data['Quantity_Dispensed']

    xgb_model = xgb.XGBRegressor(
        n_estimators=50,
        max_depth=3,
        learning_rate=0.05,
        random_state=42
    )
    xgb_model.fit(X_train, y_train)

    print("XGBoost Regressor trained successfully.")
else:
    print("Please complete Task 5 and Task 6 first to prepare training features.")
````

## Section 1: What Does a "Solution" Actually Mean?

In data science, we draw a crucial distinction between a **Model** and a **Solution**:

| Component | What It Does | Example |
| :--- | :--- | :--- |
| **Model** | Produces a statistical number | *"Projected demand for Amoxicillin next month is 11,023 units."* |
| **Solution** | Translates the prediction into an actionable decision | *"We only have 2,440 units on hand. There is an 8,583-unit deficit (Potential Shortage Risk). Submit a supplier purchase order today to avoid emergency courier fees."* |

The bridge between a raw model prediction and a business decision is called **Business Logic**.

```
[Forecasted Demand: 11,023]  ───┐
                                ├─► [Stock Gap = Demand - Stock (+8,583)] ─► [Potential Shortage Risk]
[Current On-Hand Stock: 2,440] ─┘
```

> **Clinical & Regulatory Notice**: *Our risk indicator is an analytical inventory-support tool, not a clinical directive. We use neutral terminology: "Potential Shortage Risk", "Sufficient Stock", and "Potential Excess Stock".*

````python
# --- TASK 2: YOUR CODE HERE ---
def generate_inventory_summary(medicine_id: str, forecast_demand: int, current_stock: int) -> dict:
    """
    Generates a structured inventory summary dictionary with risk classification and recommendations.
    """
    med_lookup = med[med['Medicine_ID'] == medicine_id]
    med_name = med_lookup['Medicine_Name'].values[0] if len(med_lookup) > 0 else medicine_id
    category = med_lookup['Category'].values[0] if len(med_lookup) > 0 else "General"
    
    # TODO 2.1: Calculate stock_gap
    stock_gap = forecast_demand - current_stock
    
    # TODO 2.2: Assign risk_status and natural-language recommendation
    if stock_gap > 0:
        risk_status = "Potential Shortage Risk"
        recommendation = (
            f"Forecasted demand ({forecast_demand:,} units) exceeds on-hand stock ({current_stock:,} units) "
            f"by {stock_gap:,} units. The pharmacy team may wish to review procurement plans and "
            f"schedule supplier delivery to prevent emergency orders and patient treatment delays."
        )
    elif current_stock > 2.5 * forecast_demand:
        risk_status = "Potential Excess Stock"
        recommendation = (
            f"Current stock ({current_stock:,} units) substantially exceeds projected demand ({forecast_demand:,} units). "
            f"The pharmacy team may wish to review shelf life and consider inter-hospital stock reallocation "
            f"to prevent medicine waste and expiration."
        )
    else:
        risk_status = "Sufficient Stock"
        recommendation = (
            f"Current on-hand stock ({current_stock:,} units) is well-aligned with projected demand "
            f"({forecast_demand:,} units) with a healthy reserve buffer of {abs(stock_gap):,} units. "
            f"Maintain standard reordering schedules."
        )
        
    return {
        "medicine_id": medicine_id,
        "medicine_name": med_name,
        "category": category,
        "forecasted_demand": forecast_demand,
        "current_stock": current_stock,
        "stock_gap": stock_gap,
        "risk_status": risk_status,
        "recommendation": recommendation
    }

# Test your function
test_summary = generate_inventory_summary("M001", 11023, 2440) if callable(generate_inventory_summary) else None
if test_summary:
    print("Inventory Summary Output:")
    for k, v in test_summary.items():
        print(f"  {k:18}: {v}")

# Check Your Work
gu.check_day3_task_02(generate_inventory_summary)
````

````python
# --- TASK 3: YOUR CODE HERE ---
def create_forecast_chart(ts_data: pd.DataFrame, forecast_demand: int, current_stock: int, medicine_title: str) -> go.Figure:
    """
    Generates an interactive Plotly chart showing historical demand, forecast point, and stock benchmark.
    """
    fig = go.Figure()
    
    # 1. Historical Demand Trace
    fig.add_trace(go.Scatter(
        x=ts_data.index,
        y=ts_data['Quantity_Dispensed'],
        mode='lines+markers',
        name='Historical Demand',
        line=dict(color='#1f77b4', width=2.5),
        marker=dict(size=4)
    ))
    
    # TODO 3.1: Add next month forecast point (highlighted marker)
    next_date = ts_data.index[-1] + pd.DateOffset(months=1)
    fig.add_trace(go.Scatter(
        x=[next_date],
        y=[forecast_demand],
        mode='markers',
        name='Projected Next Month',
        marker=dict(color='#ff7f0e', size=12, symbol='star')
    ))
    
    # TODO 3.2: Add current stock reference line
    fig.add_trace(go.Scatter(
        x=[ts_data.index[-6], next_date],
        y=[current_stock, current_stock],
        mode='lines',
        name=f'Current Stock ({current_stock:,})',
        line=dict(color='#2ca02c', width=2, dash='dash')
    ))
    
    fig.update_layout(
        title=f"Demand Trajectory & Stock Evaluation: {medicine_title}",
        xaxis_title="Date",
        yaxis_title="Monthly Units Dispensed",
        template="plotly_white",
        hovermode="x unified",
        margin=dict(l=40, r=40, t=50, b=40),
        legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
    )
    return fig

# Test your chart on Amoxicillin 500mg (M001)
m1_sub = dem[dem['Medicine_ID'] == 'M001'].groupby('Date').agg({'Quantity_Dispensed': 'sum'}).asfreq('MS')
fig_sample = create_forecast_chart(m1_sub, 11023, 2440, "Amoxicillin 500mg (Antibiotic)") if callable(create_forecast_chart) else None
if fig_sample:
    fig_sample.show()

# Check Your Work
gu.check_day3_task_03(fig_sample)
````

## Extension 1 — Benchmark Against a Seasonal-Naïve Baseline

A sophisticated model should earn its complexity. A strong time-series workflow therefore compares SARIMAX and XGBoost against a simple baseline.

For monthly medicine demand, a useful baseline is:

> **Next demand ≈ demand in the same month one year earlier**

If a complex model cannot beat this baseline, that is an important finding rather than something to hide.


````python
# Seasonal-naive baseline for the existing M001 test period
from sklearn.metrics import mean_absolute_error, mean_squared_error

baseline_eval = test_data[['Quantity_Dispensed', 'Demand_Lag_12']].dropna().copy()
baseline_eval['Seasonal_Naive'] = baseline_eval['Demand_Lag_12']

baseline_mae = mean_absolute_error(
    baseline_eval['Quantity_Dispensed'], baseline_eval['Seasonal_Naive']
)
baseline_rmse = np.sqrt(mean_squared_error(
    baseline_eval['Quantity_Dispensed'], baseline_eval['Seasonal_Naive']
))
baseline_mape = (
    np.abs(
        (baseline_eval['Quantity_Dispensed'] - baseline_eval['Seasonal_Naive'])
        / baseline_eval['Quantity_Dispensed'].replace(0, np.nan)
    ).mean() * 100
)

# Recalculate the two supplied model metrics on aligned test observations
sarimax_aligned = pd.Series(np.asarray(sarimax_forecast).reshape(-1), index=test_data.index[:len(sarimax_forecast)])
xgb_aligned = pd.Series(np.asarray(xgb_forecast).reshape(-1), index=test_data.index[:len(xgb_forecast)])

def metric_row(name, actual, pred):
    aligned = pd.concat([pd.Series(actual), pd.Series(pred)], axis=1).dropna()
    y, p = aligned.iloc[:, 0], aligned.iloc[:, 1]
    return {
        "Model": name,
        "MAE": mean_absolute_error(y, p),
        "RMSE": np.sqrt(mean_squared_error(y, p)),
        "MAPE (%)": np.abs((y - p) / y.replace(0, np.nan)).mean() * 100
    }

comparison_ext = pd.DataFrame([
    {
        "Model": "Seasonal Naive (12-month lag)",
        "MAE": baseline_mae,
        "RMSE": baseline_rmse,
        "MAPE (%)": baseline_mape
    },
    metric_row("SARIMAX", test_data['Quantity_Dispensed'], sarimax_aligned),
    metric_row("XGBoost", test_data['Quantity_Dispensed'], xgb_aligned)
]).sort_values("MAE").reset_index(drop=True)

print("MODEL BENCHMARK — lower is better")
display(comparison_ext.round(2))

best_model_ext = comparison_ext.iloc[0]['Model']
print(f"\nBest model by MAE: {best_model_ext}")
````

## Extension 2 — Trust-Wide Medicine Risk & Financial Exposure

Forecast accuracy matters, but pharmacy teams ultimately need to know **where to act**.

For every medicine, this section creates a next-month demand estimate using a transparent seasonal baseline, compares it with the latest recorded stock, and estimates:

- projected shortage units;
- projected excess units;
- approximate shortage exposure (£);
- approximate excess-stock value (£);
- a priority score for operational review.

The £ figures are **inventory-value proxies**, not estimates of patient harm or true NHS financial loss.


````python
# Build a trust-wide next-month inventory risk table
demand_monthly = (
    dem.groupby(['Medicine_ID', 'Date'], as_index=False)
       .agg(
           Quantity_Dispensed=('Quantity_Dispensed', 'sum'),
           Current_Stock=('Current_Stock', 'sum'),
           Unit_Cost=('Unit_Cost', 'mean'),
           Emergency_Orders=('Emergency_Orders', 'sum'),
           Medicine_Waste=('Medicine_Waste', 'sum')
       )
       .sort_values(['Medicine_ID', 'Date'])
)

risk_rows = []
for med_id, g in demand_monthly.groupby('Medicine_ID'):
    g = g.sort_values('Date').reset_index(drop=True)
    latest = g.iloc[-1]

    # Transparent seasonal forecast: same month last year where available
    forecast = float(g.iloc[-12]['Quantity_Dispensed']) if len(g) >= 12 else float(g['Quantity_Dispensed'].tail(3).mean())
    current_stock = float(latest['Current_Stock'])
    unit_cost = float(latest['Unit_Cost'])

    shortage_units = max(forecast - current_stock, 0)
    excess_units = max(current_stock - forecast, 0)
    shortage_value = shortage_units * unit_cost
    excess_value = excess_units * unit_cost

    if forecast > 0:
        coverage_ratio = current_stock / forecast
    else:
        coverage_ratio = np.nan

    if coverage_ratio < 0.50:
        risk = "Critical Shortage Risk"
        severity = 4
    elif coverage_ratio < 1.00:
        risk = "Shortage Risk"
        severity = 3
    elif coverage_ratio > 2.50:
        risk = "Excess Stock Risk"
        severity = 2
    else:
        risk = "Healthy / Monitor"
        severity = 1

    priority_score = severity * 100 + np.log1p(shortage_value + excess_value)

    risk_rows.append({
        'Medicine_ID': med_id,
        'Forecast_Next_Month': round(forecast),
        'Current_Stock': round(current_stock),
        'Coverage_Ratio': coverage_ratio,
        'Shortage_Units': round(shortage_units),
        'Excess_Units': round(excess_units),
        'Unit_Cost_GBP': unit_cost,
        'Shortage_Exposure_GBP': shortage_value,
        'Excess_Stock_Value_GBP': excess_value,
        'Risk_Status': risk,
        'Priority_Score': priority_score
    })

portfolio_risk = pd.DataFrame(risk_rows).merge(
    med[['Medicine_ID', 'Medicine_Name', 'Category']], on='Medicine_ID', how='left'
)

portfolio_risk['Total_Value_At_Risk_GBP'] = (
    portfolio_risk['Shortage_Exposure_GBP'] + portfolio_risk['Excess_Stock_Value_GBP']
)

portfolio_risk = portfolio_risk.sort_values(
    ['Priority_Score', 'Total_Value_At_Risk_GBP'], ascending=False
).reset_index(drop=True)

print("TOP 15 MEDICINES FOR OPERATIONAL REVIEW")
display(
    portfolio_risk[
        ['Medicine_ID', 'Medicine_Name', 'Category', 'Risk_Status',
         'Forecast_Next_Month', 'Current_Stock', 'Coverage_Ratio',
         'Shortage_Units', 'Excess_Units', 'Total_Value_At_Risk_GBP']
    ].head(15).style.format({
        'Coverage_Ratio': '{:.2f}x',
        'Total_Value_At_Risk_GBP': '£{:,.2f}'
    })
)

print("\nPORTFOLIO SUMMARY")
print(f"Medicines assessed: {len(portfolio_risk):,}")
print(f"Critical shortage risks: {(portfolio_risk['Risk_Status'] == 'Critical Shortage Risk').sum():,}")
print(f"Other shortage risks: {(portfolio_risk['Risk_Status'] == 'Shortage Risk').sum():,}")
print(f"Excess stock risks: {(portfolio_risk['Risk_Status'] == 'Excess Stock Risk').sum():,}")
print(f"Approx. inventory value at risk: £{portfolio_risk['Total_Value_At_Risk_GBP'].sum():,.2f}")
````

## Extension 4 — Scenario Simulator

A forecast is based on assumptions about the future. Decision-makers often need to ask **“what if?”**

The simulator below applies a simple, clearly labelled sensitivity assumption:

- a 10% rise in hospital pressure produces a 5% uplift in medicine demand;
- a 20% rise produces a 10% uplift, etc.

This **0.5 elasticity is illustrative**, not clinically validated. Its purpose is to demonstrate scenario planning and communicate uncertainty rather than claim a causal relationship.


````python
def simulate_pressure_scenario(medicine_id, hospital_pressure_change_pct=0, elasticity=0.5):
    row = portfolio_risk.loc[portfolio_risk['Medicine_ID'] == medicine_id]
    if row.empty:
        raise ValueError("Unknown Medicine_ID")
    row = row.iloc[0]

    base_forecast = float(row['Forecast_Next_Month'])
    stock = float(row['Current_Stock'])
    unit_cost = float(row['Unit_Cost_GBP'])

    scenario_forecast = base_forecast * (1 + (hospital_pressure_change_pct / 100) * elasticity)
    scenario_gap = scenario_forecast - stock
    recommended_order = max(np.ceil(scenario_gap), 0)

    return pd.DataFrame([{
        'Medicine_ID': medicine_id,
        'Medicine_Name': row['Medicine_Name'],
        'Hospital_Pressure_Change': f"{hospital_pressure_change_pct:+.0f}%",
        'Base_Forecast': round(base_forecast),
        'Scenario_Forecast': round(scenario_forecast),
        'Current_Stock': round(stock),
        'Recommended_Order_Units': int(recommended_order),
        'Indicative_Order_Value_GBP': recommended_order * unit_cost
    }])

# Example: stress-test Amoxicillin under several demand-pressure assumptions
scenario_table = pd.concat([
    simulate_pressure_scenario('M001', change)
    for change in [-20, -10, 0, 10, 20, 30]
], ignore_index=True)

display(scenario_table.style.format({
    'Indicative_Order_Value_GBP': '£{:,.2f}'
}))
````

## Extension 6 — Actionable Recommendations & Governance

### Recommended operational actions
1. **Review critical shortage medicines first.** Prioritise items where forecast demand materially exceeds current stock, especially higher-value medicines.
2. **Use model outputs as decision support, not automatic purchasing instructions.** Pharmacy staff should validate supplier lead times, pack sizes, shelf life, clinical criticality and existing purchase orders.
3. **Investigate excess-stock medicines for redistribution opportunities.** Where appropriate, stock could be reviewed across facilities before additional procurement.
4. **Monitor forecast error every month.** Compare forecasts with actual dispensing and retrain/recalibrate models when performance deteriorates.
5. **Retain a simple baseline.** A production model should continue to be compared with seasonal-naïve forecasting so additional complexity remains justified.

### Limitations
- The Northbridge scenario and datasets are part of a simulated internship project.
- The financial analysis represents **inventory value exposure**, not verified cost savings.
- The scenario elasticity is illustrative and should be estimated from real operational evidence before deployment.
- Medicine criticality, supplier lead time, shelf life, minimum order quantities and service-level targets would be needed for a production-grade replenishment policy.
- Forecasting should support qualified pharmacy/procurement professionals rather than replace clinical or operational judgement.

### What this extension demonstrates
**Forecasting → benchmarking → explainability → risk prioritisation → financial translation → scenario analysis → decision support.**

That end-to-end chain is the key portfolio contribution beyond the original internship tasks.
